```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_iris

# Load data
iris = load_iris()

df = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)

df["species"] = pd.Categorical.from_codes(
    iris.target,
    iris.target_names
)

# Basic information
print(df.head())
print(df.info())
print(df.describe())

# Missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Correlation
print("\nCorrelation:")
print(df.iloc[:, :-1].corr())

# Histograms
df.iloc[:, :-1].hist(figsize=(10, 7))
plt.tight_layout()
plt.show()

# Box plots
plt.figure(figsize=(10, 6))
sns.boxplot(data=df.iloc[:, :-1])
plt.title("Box Plot of Iris Features")
plt.show()

# Correlation heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(
    df.iloc[:, :-1].corr(),
    annot=True,
    cmap="coolwarm"
)
plt.title("Feature Correlation")
plt.show()

# Scatter plot
plt.figure(figsize=(8, 6))
sns.scatterplot(
    data=df,
    x="petal length (cm)",
    y="petal width (cm)",
    hue="species"
)
plt.title("Petal Length vs Petal Width")
plt.show()

# Pair plot
sns.pairplot(df, hue="species")
plt.show()
```
